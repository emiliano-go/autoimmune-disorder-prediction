No duplicate cols, even though the name is similar. Weird, but good ig. Lets check the target.

```python
def imbalance_report(series):
    counts = series.value_counts()
    probs = counts / counts.sum()
    entropy = -np.sum(probs * np.log2(probs))
    return {
        "num_classes": len(counts),
        "majority": f"{counts.idxmax()} ({counts.max()})",
        "minority": f"{counts.idxmin()} ({counts.min()})",
        "imbalance_ratio": round(counts.max() / counts.min(), 2),
        "cv": round(counts.std() / counts.mean(), 2),
        "shannon_evenness": round(entropy / np.log2(len(counts)), 4),
    }

for k, v in imbalance_report(df["Diagnosis"]).items():
    print(f"{k}: {v}")
```

Clearly, there are more Normal rows than any other, so a pretty clear imbalance.

```python
# Identificar tipos
cat_cols = df.select_dtypes(include=['str']).columns.tolist()
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print(f"Categorical: {len(cat_cols)}")
print(f"Numerical: {len(num_cols)}")
```

Only two categorical columns: Gender and Diagnosis. Clinical symptoms are binary, same as for antibodies. Lets check for missing data.

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print(missing)
print(f"Columns with missing values: {len(missing)}")
```

The 2500 missing values per column match exactly the Normal class count. This means these 7 antibody columns (Anti-dsDNA, Anti-Sm, Rheumatoid factor, ACPA, Anti-TPO, Anti-Tg, Anti-SMA) are NaN for all healthy patients, likely because healthy people don't get antibody tests. This is systematic missingness, not random. We want to fill those with 0 (absence of antibody = negative) before modeling.

```python
antibody_cols = ["Anti-dsDNA", "Anti-Sm", "Rheumatoid factor", "ACPA",
                 "Anti-TPO", "Anti-Tg", "Anti-SMA"]
df[antibody_cols] = df[antibody_cols].fillna(0)

df.to_parquet("../data/processed/autoimmune_cleaned.parquet", index=False)
```
